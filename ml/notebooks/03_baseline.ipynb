# 03 — Baseline XGBoost (RM1 — Skenario E0)
Melatih `XGBClassifier(objective='multi:softprob', num_class=3, tree_method='hist', random_state=42)` dengan konfigurasi default tanpa penanganan ketidakseimbangan kelas.

```python
import json
from pathlib import Path
import pandas as pd
from diaxplain_ml.features.codebook import FEATURE_ORDER, TARGET_COLUMN
from diaxplain_ml.training.experiments import run_single_experiment_cv_and_test

ML_ROOT = Path('..').resolve()
train_df = pd.read_parquet(ML_ROOT / 'data' / 'processed' / 'train.parquet')
test_df = pd.read_parquet(ML_ROOT / 'data' / 'processed' / 'test.parquet')
split_meta = json.loads((ML_ROOT / 'data' / 'processed' / 'split_indices.json').read_text())

res_e0, model_e0 = run_single_experiment_cv_and_test(
    code='E0', name='Baseline XGBoost (Default)', technique='baseline_default',
    imbalance_strategy='none',
    X_train=train_df[FEATURE_ORDER].to_numpy(), y_train=train_df[TARGET_COLUMN].to_numpy(),
    X_test=test_df[FEATURE_ORDER].to_numpy(), y_test=test_df[TARGET_COLUMN].to_numpy(),
    cv_folds_indices=split_meta['cv_folds'],
    exp_dir=ML_ROOT / 'artifacts' / 'experiments' / 'E0_baseline'
)
res_e0['test_metrics']
```
