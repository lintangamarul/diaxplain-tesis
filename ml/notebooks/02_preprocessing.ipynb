# 02 — Preprocessing & Stratified Split (Anti-Leakage)
Melakukan pembagian `train.parquet` (80%) dan `test.parquet` (20%) secara terstratifikasi (`random_state=42`) serta lipatan 5-Fold Cross-Validation di dalam data latih.

```python
from pathlib import Path
from diaxplain_ml.data.loader import ensure_dataset_available, load_raw_dataset
from diaxplain_ml.data.split import create_stratified_split, verify_no_split_leakage

ML_ROOT = Path('..').resolve()
csv_path, _ = ensure_dataset_available(ML_ROOT / 'data' / 'raw')
df, _ = load_raw_dataset(csv_path)
split_meta = create_stratified_split(df, ML_ROOT / 'data' / 'processed', test_size=0.2, n_splits=5, seed=42)
assert verify_no_split_leakage(split_meta['train_indices'], split_meta['test_indices'])
print('Train rows:', split_meta['train_rows'], 'Test rows:', split_meta['test_rows'])
```
